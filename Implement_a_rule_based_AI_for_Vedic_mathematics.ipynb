
# **Implement a rule-based AI for Vedic mathematics that includes Python functions for Urdhva Tiryagbhyam, Nikhilam Sutra, and Yavadunam methods, along with an interactive user interface to select a sutra and provide numerical inputs, and demonstrate its functionality with example operations.**

```python
def urdhva_tiryagbhyam(num1, num2):
    s_num1 = str(num1)
    s_num2 = str(num2)

    max_len = max(len(s_num1), len(s_num2))
    s_num1 = s_num1.zfill(max_len)
    s_num2 = s_num2.zfill(max_len)

    result_digits = []
    carry = 0

    for i in range(max_len * 2 - 1):
        temp_sum = 0
        for j in range(max_len):
            k = i - j
            if 0 <= k < max_len:
                digit1 = int(s_num1[j])
                digit2 = int(s_num2[k])
                temp_sum += digit1 * digit2

        temp_sum += carry
        result_digits.append(str(temp_sum % 10))
        carry = temp_sum // 10

    if carry > 0:
        result_digits.append(str(carry))

    result = int("".join(result_digits[::-1]))
    return result

print(f"32 * 41 = {urdhva_tiryagbhyam(32, 41)}")
print(f"123 * 456 = {urdhva_tiryagbhyam(123, 456)}")
print(f"9 * 8 = {urdhva_tiryagbhyam(9, 8)}")
print(f"1 * 10 = {urdhva_tiryagbhyam(1, 10)}")
print(f"0 * 123 = {urdhva_tiryagbhyam(0, 123)}")
```

```python
def nikhilam_sutra(num1, num2):
    base = 1
    while base <= num1 or base <= num2:
        base *= 10

    deviation1 = num1 - base
    deviation2 = num2 - base

    rhs = deviation1 * deviation2

    lhs = num1 + deviation2

    num_zeros_in_base = len(str(base)) - 1

    carry = 0
    if rhs < 0:
        carry = -(-(rhs) // base)
        lhs -= carry
        rhs = (carry * base) + rhs

    if rhs >= base:
        carry = rhs // base
        rhs = rhs % base
        lhs += carry

    rhs_str = str(rhs).zfill(num_zeros_in_base)

    final_product_str = str(lhs) + rhs_str
    return int(final_product_str)

print(f"9 * 8 = {nikhilam_sutra(9, 8)}")
print(f"12 * 13 = {nikhilam_sutra(12, 13)}")
print(f"97 * 98 = {nikhilam_sutra(97, 98)}")
print(f"102 * 103 = {nikhilam_sutra(102, 103)}")
print(f"98 * 103 = {nikhilam_sutra(98, 103)}")
print(f"19 * 21 = {nikhilam_sutra(19, 21)}")

print(f"4 * 5 = {nikhilam_sutra(4, 5)}")
print(f"1 * 2 = {nikhilam_sutra(1, 2)}")
print(f"999 * 998 = {nikhilam_sutra(999, 998)}")
print(f"1001 * 1002 = {nikhilam_sutra(1001, 1002)}")
print(f"997 * 1003 = {nikhilam_sutra(997, 1003)}")
```

```python
def yavadunam_sutra(num):
    base = 1
    while base <= num:
        base *= 10

    deviation = num - base

    lhs = num + deviation

    rhs = deviation ** 2

    num_zeros_in_base = len(str(base)) - 1

    carry = 0
    if rhs >= base:
        carry = rhs // base
        rhs = rhs % base
        lhs += carry

    rhs_str = str(rhs).zfill(num_zeros_in_base)

    final_product_str = str(lhs) + rhs_str

    return int(final_product_str)

print(f"8^2 = {yavadunam_sutra(8)}")
print(f"12^2 = {yavadunam_sutra(12)}")
print(f"98^2 = {yavadunam_sutra(98)}")
print(f"102^2 = {yavadunam_sutra(102)}")

def yavadunam_sutra_refined(num):
    num_str_len = len(str(num))
    base = 10**(num_str_len - 1) if num_str_len > 1 else 10

    if num < 10:
        base = 10
    elif num < 100:
        if abs(num - 10) <= abs(num - 100):
            base = 10
        else:
            base = 100
    elif num < 1000:
        if abs(num - 100) <= abs(num - 1000):
            base = 100
        else:
            base = 1000
    else:
        power = len(str(num)) - 1
        lower_base = 10**power
        upper_base = 10**(power + 1)
        if abs(num - lower_base) <= abs(num - upper_base):
            base = lower_base
        else:
            base = upper_base

    deviation = num - base
    lhs = num + deviation
    rhs = deviation ** 2

    num_zeros_in_base = len(str(base)) - 1 if base > 1 else 1 # Ensure num_zeros_in_base is at least 1

    carry = 0
    if rhs >= base:
        carry = rhs // base
        rhs = rhs % base
        lhs += carry

    rhs_str = str(rhs).zfill(num_zeros_in_base)

    if lhs == 0 and rhs == 0:
        return 0

    final_product_str = str(lhs) + rhs_str
    return int(final_product_str)

print("\n--- Refined Yavadunam Function Tests ---")
print(f"8^2 = {yavadunam_sutra_refined(8)}")
print(f"12^2 = {yavadunam_sutra_refined(12)}")
print(f"98^2 = {yavadunam_sutra_refined(98)}")
print(f"102^2 = {yavadunam_sutra_refined(102)}")
print(f"997^2 = {yavadunam_sutra_refined(997)}")
print(f"1003^2 = {yavadunam_sutra_refined(1003)}")
print(f"50^2 = {yavadunam_sutra_refined(50)}")
print(f"49^2 = {yavadunam_sutra_refined(49)}")
print(f"0^2 = {yavadunam_sutra_refined(0)}")
print(f"1^2 = {yavadunam_sutra_refined(1)}")
```

```python
def get_user_input():
    print("\n--- Vedic Mathematics AI ---")
    print("Select a Sutra for calculation:")
    print("1. Urdhva Tiryagbhyam (Multiplication of two numbers)")
    print("2. Nikhilam Sutra (Multiplication of two numbers near a base)")
    print("3. Yavadunam (Squaring of a single number)")

    sutra_choice = None
    while sutra_choice not in ['1', '2', '3']:
        sutra_choice = input("Enter your choice (1, 2, or 3): ").strip()
        if sutra_choice not in ['1', '2', '3']:
            print("Invalid choice. Please enter 1, 2, or 3.")

    numbers = []
    if sutra_choice == '1' or sutra_choice == '2':
        while True:
            try:
                num1 = int(input("Enter the first number: "))
                break
            except ValueError:
                print("Invalid input. Please enter an integer.")
        while True:
            try:
                num2 = int(input("Enter the second number: "))
                break
            except ValueError:
                print("Invalid input. Please enter an integer.")
        numbers.append(num1)
        numbers.append(num2)
    elif sutra_choice == '3':
        while True:
            try:
                num = int(input("Enter the number to square: "))
                break
            except ValueError:
                print("Invalid input. Please enter an integer.")
        numbers.append(num)

    sutra_name = {
        '1': 'Urdhva Tiryagbhyam',
        '2': 'Nikhilam Sutra',
        '3': 'Yavadunam'
    }[sutra_choice]

    return sutra_name, numbers
```

```python
def vedic_math_ai():
    while True:
        sutra_name, input_numbers = get_user_input()
        result = None

        if sutra_name == 'Urdhva Tiryagbhyam':
            if len(input_numbers) == 2:
                result = urdhva_tiryagbhyam(input_numbers[0], input_numbers[1])
                print(f"\nResult of {sutra_name} for {input_numbers[0]} * {input_numbers[1]} = {result}")
            else:
                print("Error: Urdhva Tiryagbhyam requires two numbers.")
        elif sutra_name == 'Nikhilam Sutra':
            if len(input_numbers) == 2:
                result = nikhilam_sutra(input_numbers[0], input_numbers[1])
                print(f"\nResult of {sutra_name} for {input_numbers[0]} * {input_numbers[1]} = {result}")
            else:
                print("Error: Nikhilam Sutra requires two numbers.")
        elif sutra_name == 'Yavadunam':
            if len(input_numbers) == 1:
                result = yavadunam_sutra_refined(input_numbers[0])
                print(f"\nResult of {sutra_name} for {input_numbers[0]}^2 = {result}")
            else:
                print("Error: Yavadunam requires one number.")

        print("\n")
        cont = input("Do you want to perform another calculation? (yes/no): ").strip().lower()
        if cont != 'yes':
            print("Exiting Vedic Mathematics AI. Goodbye!")
            break

vedic_math_ai()
```
